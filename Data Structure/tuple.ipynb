Explanation:

Tuples are ordered collection of item that are immutable. They are similar to list, but their immutability makes them different


```python
## creating a tuples
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
## immutable nature of tuples
## Tuples are immutable, meaning their elements cannot be changed once assigned.

lst=[1,2,3,4,5]
print(lst)

lst[1]="rahul"
print(lst)
```

```python
## Tuple Methods

print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing Tuple
Packed_tuple=1,"hello",2.14
print(Packed_tuple)
```

```python
## Unpacking tuple
a,b,c=Packed_tuple

print(a)
print(b)
print(c)
```

```python
## concept with oack and unpack

numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuple
## nested list

lst=[[1,2,3,4],[6,7,8,9],[1,"hello",3.14,"c"]]
lst[0][0:3]
```

```python
lst=[[1,2,3,4],[6,7,8,9],(1,"hello",3.14,"c")]
lst[2][0:3]
```

```python
nested_tuple = ((1,2,3),("a","b","c"),(True,False))
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()    
```
